# tkh-phase-2-project - Financial Fraud Detection - Transform notebook

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('../data/PS_20174392719_1491204439457_log.csv')
```

```python
df.info()
```

```python
# Make column names more easily readable
df_columns_fixed = df.copy()
df_columns_fixed.columns = [
    'step',
    'type',
    'amount',
    'name_orig',
    'old_balance_orig',
    'new_balance_orig',
    'name_dest',
    'old_balance_dest',
    'new_balance_dest',
    'is_fraud',
    'is_flagged_fraud'
]
df_columns_fixed
```

```python
# Dummy encode the `type` feature
df_dummies = df_columns_fixed.copy()
dummies = pd.get_dummies(df_dummies['type'], drop_first=True)
dummies_col = [c.lower() for c in dummies.columns.to_list()]
df_dummies[dummies_col] = dummies
df_dummies.drop('type', axis=1, inplace=True)
df_dummies
```

```python
# Drop extaneous features from dataframe
df_drop_extran = df_dummies.copy()
df_drop_extran.drop(['step', 'is_flagged_fraud', 'old_balance_orig', 'old_balance_dest', 'name_orig', 'name_dest'], axis=1, inplace=True)
df_drop_extran.sort_index(axis=1)
df_drop_extran
```

```python
# Save dataframe to csv file
filename = 'transform_data.csv'
df_drop_extran.to_csv(f'../data/{filename}', index=False)
```
